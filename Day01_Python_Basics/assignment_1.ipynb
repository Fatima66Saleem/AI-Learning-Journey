```python
# The user enters a string containing a number (e.g., "45"). Convert it to: an integer, a float, a string again
# Print all three values with their types
num1 = input("Enter num1: ")

as_int = int(num1)
as_float = float(num1)
as_str = str(num1)

print("Integer:", as_int, "| Type:", type(as_int))
print("Float:", as_float, "| Type:", type(as_float))
print("String:", as_str, "| Type:", type(as_str))
```

```python
# Evaluate and print the result of the following expression:
# x = 10 + 3 * 2 ** 2
# Based on what you learnt in the lecture explain why the output is what it is.

x = 10 + 3 * 2 ** 2
print(x)
# Python operator precedence follows a specific order: exponent (**) has the highest priority, followed by multiplication (*) and division (/), and then addition (+) and subtraction (-) last. In this expression, Python first calculates 2 ** 2 = 4, then multiplies 3 * 4 = 12, and finally adds 10 + 12 = 22. That's why the output is 22, not something else if we calculated left to right without precedence
```

```python
# Write a program to swap values of two numbers entered by the user

num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2: "))

temp = num1
num1 = num2
num2 = temp
print("After Swapping")
print("The value of num1 is: ",num1)
print("The value of num2 is: ",num2)
```

```python
# Ask the user for a temperature in Celsius (string input). Convert it to a float
# then calculate and print temperature in Fahrenheit.
# Conversion formula: FahrenheitTemp = (CelsiusTemp * (9/5)) + 32

CelsiusTemp = input("Enter the temperature in Celsius: ")
float_temp = float(CelsiusTemp)

fahrenheit_temp = (float_temp * (9/5)) + 32
print(f"Fahrenheit Temperature is {fahrenheit_temp}")
```

```python
# Take the radius (r) as user input and print the area.
# Use the formula: Area = π * r (value of π = 3.14)

r = float(input("Enter the value of radius: "))

area = 3.14 * r**2

print("The area is: ", area)
```

```python
# Ask the user for: Principal (P), Rate (R), Time (T). Convert all to float and
# compute simple interest: SI = (P ∗ R ∗ T)/100

P = float(input("Enter the value of Principal: "))
R = float(input("Enter the value of Rate: "))
T = float(input("Enter the value of Time: "))

SI = (P * R * T)/100
print("Interest is: ", SI)
```

```python
# Take a decimal number as input (like 45.78 ) and output its:
# • integer part - 45
# • fractional part - .78

num = float(input("Enter the input number as 45.78: "))

integer_part = int(num)        
fractional_part = num - integer_part

print("Integer part:", integer_part)
print(f"Fractional part: {fractional_part:.2f}")
```
