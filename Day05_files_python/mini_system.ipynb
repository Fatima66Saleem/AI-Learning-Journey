```python


while(True):
    print("1. Generate social media caption")
    print("2. Analyze customer reviews")
    print("3. Highlight product details")
    print("4. Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
        f = open("product_descriptions.txt", "r")
        content = f.read()
        f.close()
        prompt = f"Write a short, catchy Instagram caption with emojis for each of the following products: {content}"
        response = get_llm_response(prompt)
        display(Markdown(response))

    elif choice == "2":
        files = ["reviews_keychain.txt", "reviews_mug.txt", "reviews_wallhanging.txt", "customer_reviews.txt"]
        for file in files:
            f = open(file, "r")
            reviews = f.read()
            f.close()

            prompt = f"Classify the following customer reviews as either Positive or Negative overall, and briefly explain why in one line: {reviews}"
            response = get_llm_response(prompt)

            print(file, "->")
            display(Markdown(response))

    elif choice == "3":
        f = open("product_descriptions.txt", "r")
        content = f.read()
        f.close()

        prompt = f"Given the following product descriptions, highlight the material or product type (like resin, wood, ceramic) in orange, and highlight the occasion or use (like gift, birthday, home decor) in green. Provide the output as HTML suitable for display in a Jupyter notebook: {content}"
        response = get_llm_response(prompt)
        display(HTML(response))

    elif choice == "4":
        print("Exiting the system.....")
        break

    else:
        print("Invalid choice, please try again.")
```
