```python
# instance method

class Laptop:
    storage_type = "ssd"
    def __init__(self, type):
        self.type = type

    def info(self): # instance method can access both instance and class attributes
        print(f"Laptop has {self.type} RAM and {self.storage_type} of storage")

l1 = Laptop("type name")

l1.info()
```
