```python
class Laptop:
    storage_type = "ssd"
    def __init__(self, type):
        self.type = type

    @classmethod # decortor that chnage the behavior of a method and make is class method
    def get_storage(cls):
        print(f"Laptop has {cls.storage_type} storage")

    def info(self): # instance method can access both instance and class attributes
        print(f"Laptop has {self.type} RAM and {self.storage_type} of storage")

    @staticmethod
    def cal_discount(price, discount):
        final_price = price - (discount * price / 100)
        print(f"Final price = {final_price}")

l1 = Laptop("type name")
l1.cal_discount(40000, 80)
```
