```python
class Student:
    uni_name = "Riphah International University"
    def __init__(self, name , cgpa , age):
        self.name = name
        self.cgpa = cgpa
        self.age = age

student1 = Student("Fatima" , 4.0 , 21)
print(Student.uni_name) # class attribute uni name
print(student1.name, student1.cgpa, student1.age) # object attributes , object can access both type of attributes but class can access only class type attributes
```
