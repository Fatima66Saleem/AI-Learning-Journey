```python
class Product:
    count = 0
    def __init__(self, name, price):
        self.name = name
        self.price = price
        Product.count += 1

    def info(self): # instance method
        print(f"The name of product is {self.product} and price is {self.price}")


    @classmethod # class method
    def track_objects(cls):
        print(f"Total objects are {cls.count}")

    @staticmethod
    def cal_discount(price, discount):
        final_price = price - (discount * price / 100)
        print(f"final price is {final_price}")

p1 = Product("Mobile", 20000)
p2 = Product("Bag" , 5000)

p1.track_objects()

p1.info()

p1.cal_discount(p1.price, 10)

        
```
