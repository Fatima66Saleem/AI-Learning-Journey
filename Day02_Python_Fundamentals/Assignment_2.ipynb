```python
# Write a program that takes salary as input. Using conditional statements, calculate the final tax rate based on these rules:
# • If salary < 30,000 → 5%
# • If salary is 30,000–70,000 → 15%
# • If salary > 70,000 → 25%

salary = float(input("Enter your salary: "))

if salary < 30000:
    tax_rate = 5
elif salary <= 70000:
    tax_rate = 15
else:
    tax_rate = 25

tax = salary * (tax_rate / 100)

print(f"Salary: {salary}")
print(f"Tax Rate: {tax_rate}%")
print(f"Tax Amount: {tax}")
```

```python
# Write a function that takes two integers and and prints all even
# numbers between them (inclusive).

def even_num(num1 , num2):
    for i in range(num1, num2+1):
        if(i % 2 == 0):
            print("Even number", i)



num1 = int(input("Enter the number 1: "))
num2 = int(input("Enter the number 2: "))
print(even_num(num1, num2))
```

```python
# Write a function that prints the digits of a number, n .
# For eg: n = 312 , there are 3 digits in it 3, 1 and 2 & we need to print them.
# [Hint - The right most digit of a number N is N%10.
# And to remove the right most digit from a number, we can do N = N / 10.]

def print_digits(n):
    while n > 0:
        digit = n % 10
        print(digit)
        n = n // 10

num = int(input("Enter a number: "))
print_digits(num) 
```

```python
# Write a function to return the count the number of digits in a number, n .
def count_num(n):
    count = 0  # counter starts at 0, increases with each digit found
    for i in str(n):  # convert n to a string so we can loop through each digit
        count += 1  # increment count for every character (digit) found
    return count  # return the final total digit count

n = int(input("Enter the number n: "))
print("The total count of n is: ", count_num(n))
```

```python
# Write a function to return the sum of digits of a number, n .

def sum_of_digits(n):
    s = 0
    while n > 0:
        digit = n % 10   # extract the rightmost digit
        s = s + digit    # add that digit to the sum
        n = n // 10      # remove the rightmost digit
    return s

n = int(input("Enter the number n: "))
print("The sum of digits is: ", sum_of_digits(n))
```

```python
# Let’s create a Simple Calculator that performs arithmetic operations. Create a function  calculator(a, b, operation) that performs 
# addition, subtraction, multiplication, or division based on the operation parameter. 
# operation parameter can have values ‘+’ , ‘-’ , '*’ & ‘/’ .
# Let's create a Simple Calculator that performs arithmetic operations.
# Create a function calculator(a, b, operation) that performs
# addition, subtraction, multiplication, or division based on the operation parameter.
# operation parameter can have values '+' , '-' , '*' & '/' .

def calculator(a, b, operation):
    if operation == '+':
        print("Addition: ", a + b)
    elif operation == '-':
        print("Subtraction: ", a - b)
    elif operation == '*':
        print("Multiplication: ", a * b)
    elif operation == '/':
        if b == 0:
            print("Error: Cannot divide by zero")
        else:
            print("Division: ", a / b)
    else:
        print("Invalid input")

a = int(input("Enter the number a: "))
b = int(input("Enter the number b: "))
operation = input("Enter the operation + , - , * , / : ")
calculator(a, b, operation)
```

```python
# Write a function that returns True if n is a prime number, and False otherwise, using a loop.

def is_prime(n):
    if n < 2:
        return False   # numbers less than 2 are not prime

    for i in range(2, n):   # check all numbers from 2 to n-1
        if n % i == 0:      # if n divides evenly by any i, it's not prime
            return False

    return True   # no divisor found, so n is prime


n = int(input("Enter a number: "))
print(is_prime(n))
```

```python
# Number Guessing Game
# Given a secret number, ask the user to guess it and print:
# "Too high" if the guess is above the number
# "Too low" if the guess is below
# "Correct!" if the guess matches

secret = 42   # secret number decided in advance

while True:
    guess = int(input("Guess the number: "))

    if guess > secret:
        print("Too high")
    elif guess < secret:
        print("Too low")
    else:
        print("Correct!")
        break   # correct guess, exit the loop
```
