```python
# for loop=> sequential traversal
string = "Hello World"
# in => membership operator (it track the members of sequencs) and it also check the presence
for var in string:
    print(var)
```

```python
word = "artifical intelligence"
# count the no of 'i' in word
count = 0
for ch in word:
    if(ch == 'i'):
       count =count + 1
print("count of i:", count)
```

```python
word = "artifical intelligence"
# count the vovels in word
count = 0
for ch in word:
    if(ch == 'a' or ch == 'e' or ch == 'i' or ch =='o' or ch == 'u'):
       count =count + 1
print("count of i:", count)
```

```python
# range function
for i in range(1, 6): # 1 is the starting value , 6 is the stop value which is n-1 
    print(i)  
```

```python
# range function
for i in range(1, 6, 2): # 1 is the starting value , 6 is the stop value which is n-1  and 2 is the step value( increment by 2)
    print(i)
```
