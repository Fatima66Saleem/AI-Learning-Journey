```python
# finite loop => 5
count = 1 #iterator
while count <= 5:
    print("Hellow World", count)
    count += 1
print("after loop, count: ", count)
```

```python
# print number from 1 to 5

num = 1
while num <= 5:
    print(num)
    num += 1
```

```python
# print number from 5 to 1 (reverse)

num = 5
while (num >= 1):
    print(num)
    num -= 1
```

```python
# multiplication table of number n 

n = int(input("Enter the value of n: "))
i = 1
while (i <= 10):
    print(n * i)
    i += 1
```

```python
# break statement 
num = 1
while num <= 10:
    if(num % 6 == 0):
        break #terminate 
    print(num)
    num += 1
print("outside loop")
```

```python
# continue
num = 1
while num <= 10:
    if(num % 3 == 0):
        num += 1
        continue # Skip the current iteration and jump on next iteration
    print(num)
    num += 1
print("outside loop")
```

```python
# print the odd numbers from 1 to 10 using continue

num = 1
while num <= 10:
    if(num % 2 == 0):
        num += 1
        continue # Skip the current iteration and jump on next iteration
    print(num)
    num += 1
print("outside loop")
```
