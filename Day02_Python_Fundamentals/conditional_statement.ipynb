```python
age = int(input("Enter age: "))

if age < 13:
    print("Child")
elif age >= 13 and age < 18: # 13-18
    print("Teenager")
else:
    print("adult")
```

```python
username = input("Enter the username: ")
password = input("Enter the Password: ")

if username == "admin" and password == "pass":
    print("login successful")
elif username != "admin":
    print("incorrect username")
elif password != "pass":
    print("incorrect password")
else:
    print("incorrect username and password")
```

```python
n = int(input("enter the value of n: "))

if n % 5 == 0:
    print("n is the multiple of 5")
else:
    print("n is not multiple of 5")
```

```python
n = int(input("enter the value of n: "))

if n % 2 == 0:
    print("n is even")
else:
    print("n is odd")
```

```python
# nested condition
username = input("Enter the username: ")
password = input("Enter the Password: ")

if username == "admin" and password == "pass":
    print("login successful")
else:
    if username != "admin":
        print("incorrect username")
    elif password != "pass":
        print("incorrect password")
    else:
        print("incorrect username and password")
```

```python
# match case conditional statement
color = input("Enter color: ")

match color:
    case "Red":
        print("Stop")
    case "Yellow":
        print("Look")
    case "Green":
        print("Go")
    case _:
        print("wrong color")
```
