```python
# Function defination
def average(a, b, c): # a b c are the parameters
    avg = (a+b+c) / 3
    return avg # return is the keyword 

result = average(6, 7, 45) # function call and values are the arguments
print(result) # result here is also the argument 
```

```python
def sum(a, b=1): # a is non default parameter and b is default parameter => default para always follows the non default para
    return a+b

print(sum(5)) # 5+1 => 6 here 1 is consider as default parameter
```

```python
sum = lambda a, b: a+b # lambda function is used for simple task and it has only one expression and return some value 

print(sum(5,6))
```

```python
def cal_fact(n):
    fact = 1
    for i in range(1, n+1):
        fact = fact * i
    return fact
 
print(cal_fact(6)) 
```
